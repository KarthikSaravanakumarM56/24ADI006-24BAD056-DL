```python
# EXPERIMENT NO: 4
print(" 24BAD056 - M.Karthik Saravanakumar")

# 1. IMPORT LIBRARIES
import os
import tarfile
import urllib.request

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input

from tensorflow.keras.layers import (
    Dense,
    GlobalAveragePooling2D
)
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam

# 2. CHECK TENSORFLOW AND GPU
print("TensorFlow Version:", tf.__version__)
gpu = tf.config.list_physical_devices("GPU")
print("GPU Available:", gpu)
if not gpu:
    print("Running on CPU - using FAST training settings.")

# 3. DOWNLOAD DATASET DIRECTLY FROM INTERNET
DATASET_URL = (
    "https://storage.googleapis.com/"
    "download.tensorflow.org/example_images/"
    "flower_photos.tgz"
)
DATASET_FILE = "/content/flower_photos.tgz"
DATASET_DIR = "/content/flower_photos"

print("\nChecking dataset...")
if not os.path.exists(DATASET_DIR):
    print("Downloading dataset...")
    urllib.request.urlretrieve(
        DATASET_URL,
        DATASET_FILE
    )
    print("Download completed.")
    print("Extracting dataset...")
    with tarfile.open(
        DATASET_FILE,
        "r:gz"
    ) as tar:
        tar.extractall(
            path="/content"
        )
    print("Dataset extracted.")
else:
    print("Dataset already exists.")

# 4. GET CLASS NAMES
class_names = sorted(
    [
        folder
        for folder in os.listdir(DATASET_DIR)
        if os.path.isdir(
            os.path.join(
                DATASET_DIR,
                folder
            )
        )
    ]
)
num_classes = len(class_names)
print("\nClasses:")
print(class_names)
print(
    "Number of Classes:",
    num_classes
)

# 5. FAST TRAINING SETTINGS
IMG_SIZE = (160, 160)
BATCH_SIZE = 32
SEED = 123
EPOCHS = 2

# 6. LOAD TRAINING DATA
print("\nLoading training data...")

train_ds = tf.keras.utils.image_dataset_from_directory(

    DATASET_DIR,

    validation_split=0.20,

    subset="training",

    seed=SEED,

    image_size=IMG_SIZE,

    batch_size=BATCH_SIZE,

    label_mode="int"
)

# 7. LOAD VALIDATION DATA
print("\nLoading validation data...")

val_ds = tf.keras.utils.image_dataset_from_directory(

    DATASET_DIR,

    validation_split=0.20,

    subset="validation",

    seed=SEED,

    image_size=IMG_SIZE,

    batch_size=BATCH_SIZE,

    label_mode="int"
)

# 8. CREATE SMALL TEST SET
# We use a small part of validation data as test data
# to keep CPU execution fast.

val_batches = tf.data.experimental.cardinality(
    val_ds
).numpy()

test_batches = max(
    1,
    int(val_batches * 0.20)
)

test_ds = val_ds.take(
    test_batches
)

val_ds = val_ds.skip(
    test_batches
)


# ============================================================
# 9. LIMIT DATASET SIZE FOR FAST TRAINING
# ============================================================

# 100 batches × 32 images = approximately 3200 images maximum.
# The original dataset has around 2936 training images,
# so this effectively uses the complete training portion.

train_ds = train_ds.take(100)

val_ds = val_ds.take(20)

test_ds = test_ds.take(20)


# ============================================================
# 10. PREFETCH DATA
# ============================================================

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(
    AUTOTUNE
)

val_ds = val_ds.prefetch(
    AUTOTUNE
)

test_ds = test_ds.prefetch(
    AUTOTUNE
)

print("\nDataset preparation completed.")

# 11. DISPLAY SAMPLE IMAGES
plt.figure(
    figsize=(9, 7)
)

for images, labels in train_ds.take(1):

    for i in range(
        min(9, len(images))
    ):

        plt.subplot(
            3,
            3,
            i + 1
        )

        plt.imshow(
            images[i].numpy().astype(
                "uint8"
            )
        )

        plt.title(
            class_names[
                labels[i].numpy()
            ]
        )

        plt.axis("off")

plt.tight_layout()

plt.show()

# 12. PREPROCESS IMAGES FOR RESNET-50
def preprocess_data(
    images,
    labels
):

    images = tf.cast(
        images,
        tf.float32
    )

    images = preprocess_input(
        images
    )

    return images, labels


train_ds = train_ds.map(
    preprocess_data,
    num_parallel_calls=AUTOTUNE
)

val_ds = val_ds.map(
    preprocess_data,
    num_parallel_calls=AUTOTUNE
)

test_ds = test_ds.map(
    preprocess_data,
    num_parallel_calls=AUTOTUNE
)

# 13. LOAD PRE-TRAINED RESNET-50
print("\nLoading ResNet-50...")

base_model = ResNet50(

    weights="imagenet",

    include_top=False,

    input_shape=(
        160,
        160,
        3
    )
)

print(
    "ResNet-50 loaded successfully."
)

# 14. FREEZE PRE-TRAINED LAYERS
base_model.trainable = False

print(
    "Pre-trained layers frozen."
)

# 15. ADD NEW CLASSIFICATION LAYER
x = base_model.output

x = GlobalAveragePooling2D()(x)

x = Dense(
    32,
    activation="relu"
)(x)

output = Dense(
    num_classes,
    activation="softmax"
)(x)


# Create final model

model = Model(

    inputs=base_model.input,

    outputs=output
)

# 16. COMPILE MODEL
model.compile(

    optimizer=Adam(
        learning_rate=0.0001
    ),

    loss=(
        "sparse_categorical_crossentropy"
    ),

    metrics=[
        "accuracy"
    ]
)


print("\nModel compiled successfully.")

# 17. DISPLAY MODEL SUMMARY
print("\nModel Summary:")

model.summary()

# 18. TRAIN MODEL
print("\n======================================")
print("STARTING TRAINING")
print("======================================")

history = model.fit(

    train_ds,

    validation_data=val_ds,

    epochs=EPOCHS
)

# 19. TEST MODEL
print("\n======================================")
print("EVALUATING MODEL")
print("======================================")

test_loss, test_accuracy = model.evaluate(
    test_ds
)


print("\n======================================")
print("FINAL RESULTS")
print("======================================")

print(
    "Test Loss     :",
    test_loss
)

print(
    "Test Accuracy :",
    test_accuracy
)

# 20. ACCURACY VS EPOCH
plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["accuracy"],
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    marker="o",
    label="Validation Accuracy"
)

plt.title(
    "Accuracy vs Epoch"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()

plt.grid(True)

plt.show()

# 21. LOSS VS EPOCH
plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["loss"],
    marker="o",
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    marker="o",
    label="Validation Loss"
)

plt.title(
    "Loss vs Epoch"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.legend()

plt.grid(True)

plt.show()

# 22. SAMPLE PREDICTIONS
print("\nGenerating sample predictions...")

plt.figure(
    figsize=(12, 8)
)

for images, labels in test_ds.take(1):

    predictions = model.predict(
        images,
        verbose=0
    )

    predicted_classes = np.argmax(
        predictions,
        axis=1
    )

    for i in range(
        min(9, len(images))
    ):

        plt.subplot(
            3,
            3,
            i + 1
        )

        # Reverse ResNet preprocessing
        image = images[i].numpy()

        image = image + np.array(
            [
                103.939,
                116.779,
                123.680
            ]
        )

        # BGR → RGB
        image = image[:, :, ::-1]

        image = np.clip(
            image / 255.0,
            0,
            1
        )

        plt.imshow(
            image
        )

        actual = class_names[
            labels[i].numpy()
        ]

        predicted = class_names[
            predicted_classes[i]
        ]

        plt.title(
            "Actual: " + actual +
            "\nPredicted: " + predicted
        )

        plt.axis("off")

plt.tight_layout()
plt.show()

# 23. SAVE MODEL
MODEL_PATH = (
    "/content/"
    "resnet50_transfer_learning_flowers.keras"
)

model.save(
    MODEL_PATH
)

print(
    "\nModel saved at:"
)

print(
    MODEL_PATH
)

# 24. FINAL OUTPUT
print("\n======================================")
print("EXPERIMENT 4 COMPLETED")
print("======================================")

print(
    "Dataset       : TensorFlow Flowers"
)

print(
    "Model         : ResNet-50"
)

print(
    "Classes       :",
    num_classes
)

print(
    "Image Size    :",
    IMG_SIZE
)

print(
    "Batch Size    :",
    BATCH_SIZE
)

print(
    "Epochs        :",
    EPOCHS
)

print(
    "Test Accuracy :",
    test_accuracy
)

print("======================================")
```
