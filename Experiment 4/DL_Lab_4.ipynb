```python
print("24BAD056 - M.Karthik Saravanakumar")

# STEP 1: INSTALL REQUIRED LIBRARIES
!pip install -q transformers

# STEP 2: IMPORT REQUIRED LIBRARIES
import os
import glob
import tarfile
import urllib.request

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input
from tensorflow.keras.layers import (
    Dense,
    GlobalAveragePooling2D,
    Dropout
)
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam

from transformers import pipeline
from PIL import Image

# STEP 3: CHECK TENSORFLOW VERSION
print("TensorFlow Version:", tf.__version__)
print(
    "GPU Available:",
    tf.config.list_physical_devices("GPU")
)

# PART A: DATASET PREPARATION
# STEP 4: DOWNLOAD IMAGE DATASET
DATASET_URL = (
    "https://storage.googleapis.com/"
    "download.tensorflow.org/example_images/"
    "flower_photos.tgz"
)
DATASET_FILE = "/content/flower_photos.tgz"
DATASET_DIR = "/content/flower_photos"

if not os.path.exists(DATASET_DIR):
    print("\nDownloading dataset...")
    urllib.request.urlretrieve(
        DATASET_URL,
        DATASET_FILE
    )
    print("Download completed.")
    print("Extracting dataset...")
    with tarfile.open(
        DATASET_FILE,
        "r:gz"
    ) as tar:
        tar.extractall(
            "/content"
        )
    print("Dataset extracted.")
else:
    print("\nDataset already exists.")

# STEP 5: GET CLASS NAMES
class_names = sorted(
    [
        folder
        for folder in os.listdir(DATASET_DIR)
        if os.path.isdir(
            os.path.join(
                DATASET_DIR,
                folder
            )
        )
    ]
)
num_classes = len(class_names)
print("\nClasses:")
print(class_names)
print(
    "Number of Classes:",
    num_classes
)

# STEP 6: DEFINE IMAGE SETTINGS
IMG_SIZE = (
    160,
    160
)
BATCH_SIZE = 32
SEED = 123

# STEP 7: LOAD TRAINING DATA
print("\nLoading training data...")
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR,
    validation_split=0.20,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int"
)

# STEP 8: LOAD VALIDATION DATA
print("\nLoading validation data...")
val_ds = tf.keras.utils.image_dataset_from_directory(
    DATASET_DIR,
    validation_split=0.20,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int"
)

# STEP 9: CREATE TEST DATA
# A small part of validation data is used as test data.
val_batches = tf.data.experimental.cardinality(
    val_ds
).numpy()
test_batches = max(
    1,
    int(val_batches * 0.20)
)
test_ds = val_ds.take(
    test_batches
)
val_ds = val_ds.skip(
    test_batches
)

# STEP 10: LIMIT DATA FOR FASTER TRAINING
train_ds = train_ds.take(100)
val_ds = val_ds.take(20)
test_ds = test_ds.take(20)

# STEP 11: PREFETCH DATA
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.prefetch(
    AUTOTUNE
)
val_ds = val_ds.prefetch(
    AUTOTUNE
)
test_ds = test_ds.prefetch(
    AUTOTUNE
)
print(
    "\nDataset preparation completed."
)

# STEP 12: DISPLAY SAMPLE IMAGES
plt.figure(
    figsize=(9, 7)
)
for images, labels in train_ds.take(1):
    for i in range(
        min(9, len(images))
    ):
        plt.subplot(
            3,
            3,
            i + 1
        )
        plt.imshow(
            images[i].numpy().astype(
                "uint8"
            )
        )
        plt.title(
            class_names[
                labels[i].numpy()
            ]
        )
        plt.axis("off")
plt.tight_layout()
plt.show()

# STEP 13: PREPROCESS IMAGES
def preprocess_data(
    images,
    labels
):
    images = tf.cast(
        images,
        tf.float32
    )
    images = preprocess_input(
        images
    )
    return images, labels

train_ds = train_ds.map(
    preprocess_data,
    num_parallel_calls=AUTOTUNE
)
val_ds = val_ds.map(
    preprocess_data,
    num_parallel_calls=AUTOTUNE
)
test_ds = test_ds.map(
    preprocess_data,
    num_parallel_calls=AUTOTUNE
)

# PART B: IMPLEMENTING TRANSFER LEARNING
# STEP 14: LOAD PRE-TRAINED RESNET-50
print("\nLoading pre-trained ResNet-50...")
base_model = ResNet50(
    weights="imagenet",
    include_top=False,
    input_shape=(
        160,
        160,
        3
    )
)
print(
    "ResNet-50 loaded successfully."
)

# STEP 15: FREEZE FEATURE EXTRACTION LAYERS
base_model.trainable = False
print(
    "Pre-trained ResNet-50 layers are frozen."
)

# STEP 16: REPLACE FINAL CLASSIFICATION LAYER
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(
    32,
    activation="relu"
)(x)
x = Dropout(
    0.3
)(x)
output = Dense(
    num_classes,
    activation="softmax"
)(x)
model = Model(
    inputs=base_model.input,
    outputs=output
)

# STEP 17: COMPILE MODEL
model.compile(
    optimizer=Adam(
        learning_rate=0.0001
    ),
    loss=(
        "sparse_categorical_crossentropy"
    ),
    metrics=[
        "accuracy"
    ]
)
print(
    "\nModel compiled successfully."
)

# STEP 18: DISPLAY MODEL SUMMARY
model.summary()

# PART C: MODEL TRAINING AND PERFORMANCE EVALUATION
# STEP 19: TRAIN TRANSFER LEARNING MODEL
print("\n====================================")
print("TRANSFER LEARNING TRAINING")
print("====================================")

history_initial = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=2
)

# STEP 20: FINE-TUNING
print("\n====================================")
print("FINE-TUNING RESNET-50")
print("====================================")

# Unfreeze ResNet-50
base_model.trainable = True

# Freeze all layers except last 5
for layer in base_model.layers[:-5]:
    layer.trainable = False
print(
    "Last 5 ResNet-50 layers are unfrozen."
)

# STEP 21: COMPILE AGAIN
model.compile(
    optimizer=Adam(
        learning_rate=0.00001
    ),
    loss=(
        "sparse_categorical_crossentropy"
    ),
    metrics=[
        "accuracy"
    ]
)

# STEP 22: FINE-TUNE MODEL
history_fine = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=1
)

# STEP 23: COMBINE TRAINING HISTORY
accuracy = (
    history_initial.history[
        "accuracy"
    ]
    +
    history_fine.history[
        "accuracy"
    ]
)
val_accuracy = (
    history_initial.history[
        "val_accuracy"
    ]
    +
    history_fine.history[
        "val_accuracy"
    ]
)
loss = (
    history_initial.history[
        "loss"
    ]
    +
    history_fine.history[
        "loss"
    ]
)
val_loss = (
    history_initial.history[
        "val_loss"
    ]
    +
    history_fine.history[
        "val_loss"
    ]
)

# STEP 24: DISPLAY RESULTS
print("\n====================================")
print("TRAINING RESULTS")
print("====================================")

print(
    "Training Accuracy:",
    round(
        accuracy[-1] * 100,
        2
    ),
    "%"
)
print(
    "Validation Accuracy:",
    round(
        val_accuracy[-1] * 100,
        2
    ),
    "%"
)
print(
    "Training Loss:",
    round(
        loss[-1],
        4
    )
)
print(
    "Validation Loss:",
    round(
        val_loss[-1],
        4
    )
)

# STEP 25: TEST MODEL
print("\n====================================")
print("MODEL TESTING")
print("====================================")

test_loss, test_accuracy = model.evaluate(
    test_ds
)
print(
    "\nTest Loss:",
    round(
        test_loss,
        4
    )
)
print(
    "Test Accuracy:",
    round(
        test_accuracy * 100,
        2
    ),
    "%"
)

# STEP 26: ACCURACY VS EPOCH
epochs_range = range(
    1,
    len(accuracy) + 1
)
plt.figure(
    figsize=(8, 5)
)
plt.plot(
    epochs_range,
    accuracy,
    marker="o",
    label="Training Accuracy"
)
plt.plot(
    epochs_range,
    val_accuracy,
    marker="o",
    label="Validation Accuracy"
)
plt.xlabel(
    "Epoch"
)
plt.ylabel(
    "Accuracy"
)
plt.title(
    "Accuracy vs Epoch"
)
plt.legend()
plt.grid(True)
plt.show()

# STEP 27: LOSS VS EPOCH
plt.figure(
    figsize=(8, 5)
)
plt.plot(
    epochs_range,
    loss,
    marker="o",
    label="Training Loss"
)
plt.plot(
    epochs_range,
    val_loss,
    marker="o",
    label="Validation Loss"
)
plt.xlabel(
    "Epoch"
)
plt.ylabel(
    "Loss"
)
plt.title(
    "Loss vs Epoch"
)
plt.legend()
plt.grid(True)
plt.show()

# STEP 28: RESNET-50 SAMPLE PREDICTION
print("\n====================================")
print("RESNET-50 IMAGE PREDICTION")
print("====================================")

for images, labels in test_ds.take(1):
    predictions = model.predict(
        images,
        verbose=0
    )
    predicted_classes = np.argmax(
        predictions,
        axis=1
    )
    plt.figure(
        figsize=(12, 8)
    )
    for i in range(
        min(9, len(images))
    ):
        plt.subplot(
            3,
            3,
            i + 1
        )
        image = images[i].numpy()
        image = image + np.array(
            [
                103.939,
                116.779,
                123.680
            ]
        )
        image = image[:, :, ::-1]
        image = np.clip(
            image / 255.0,
            0,
            1
        )
        plt.imshow(
            image
        )
        actual = class_names[
            labels[i].numpy()
        ]
        predicted = class_names[
            predicted_classes[i]
        ]
        plt.title(
            "Actual: "
            + actual
            + "\nPredicted: "
            + predicted
        )
        plt.axis("off")
    plt.tight_layout()
    plt.show()

# PART D: USING HUGGING FACE PRE-TRAINED MODEL
# STEP 29: LOAD HUGGING FACE MODEL
print("\n====================================")
print("HUGGING FACE MODEL")
print("====================================")

print(
    "\nLoading pre-trained Vision Transformer..."
)
classifier = pipeline(
    "image-classification",
    model="google/vit-base-patch16-224"
)
print(
    "Hugging Face model loaded successfully."
)

# STEP 30: GET SAMPLE IMAGE
image_files = glob.glob(
    DATASET_DIR
    + "/*/*.jpg"
)
sample_path = image_files[0]
hf_image = Image.open(
    sample_path
).convert("RGB")
print(
    "\nSample image:",
    sample_path
)
display(
    hf_image
)

# STEP 31: HUGGING FACE IMAGE CLASSIFICATION
print(
    "\nHugging Face Predictions"
)
hf_predictions = classifier(
    hf_image
)
for result in hf_predictions[:5]:
    print(
        result["label"],
        ":",
        round(
            result["score"] * 100,
            2
        ),
        "%"
    )

# STEP 32: RESNET-50 PREDICTION ON SAME IMAGE
image_array = np.array(
    hf_image
)
resnet_image = tf.image.resize(
    image_array,
    (
        160,
        160
    )
)
resnet_image = tf.cast(
    resnet_image,
    tf.float32
)
resnet_image = preprocess_input(
    resnet_image
)
resnet_image = tf.expand_dims(
    resnet_image,
    axis=0
)
resnet_prediction = model.predict(
    resnet_image,
    verbose=0
)
resnet_class = np.argmax(
    resnet_prediction[0]
)
resnet_confidence = np.max(
    resnet_prediction[0]
)

# STEP 33: COMPARE RESNET-50 AND HUGGING FACE
hf_top = hf_predictions[0]
print("\n====================================")
print("MODEL COMPARISON")
print("====================================")

print(
    "ResNet-50 Prediction:",
    class_names[
        resnet_class
    ]
)
print(
    "ResNet-50 Confidence:",
    round(
        resnet_confidence * 100,
        2
    ),
    "%"
)
print(
    "Hugging Face Prediction:",
    hf_top["label"]
)
print(
    "Hugging Face Confidence:",
    round(
        hf_top["score"] * 100,
        2
    ),
    "%"
)

# STEP 34: SAVE MODEL
MODEL_PATH = (
    "/content/"
    "resnet50_transfer_learning.keras"
)
model.save(
    MODEL_PATH
)
print(
    "\nModel saved successfully."
)

# STEP 35: FINAL OBSERVATION
print("\n====================================")
print("FINAL OBSERVATION")
print("====================================")

print(
    "1. Transfer learning using ResNet-50 "
    "was successfully implemented."
)
print(
    "2. Pre-trained ImageNet features were "
    "used for image recognition."
)
print(
    "3. The final classification layer was "
    "replaced according to the dataset classes."
)
print(
    "4. Feature extraction layers were frozen "
    "during initial training."
)
print(
    "5. The last few layers were fine-tuned "
    "using a small learning rate."
)
print(
    "6. Training and validation accuracy "
    "and loss were recorded."
)
print(
    "7. Test accuracy was calculated."
)
print(
    "8. Accuracy and loss graphs were plotted."
)
print(
    "9. A pre-trained Vision Transformer "
    "from Hugging Face was loaded."
)
print(
    "10. Image classification was performed "
    "using the Hugging Face model."
)
print(
    "11. ResNet-50 and Hugging Face predictions "
    "were compared."
)
print(
    "\nExperiment 4 completed successfully."
)
```
